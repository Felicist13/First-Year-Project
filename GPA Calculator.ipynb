```python
dic = {"A+": 5, "A": 5, "A-": 4.5, "B+": 4.0, "B": 3.5, "B-": 3.0}
```

```python
print(dic["A"])
```

```python
def calculate(grades):
    result = 0
    for i in grades:
        result += dic[i]
    result = result/len(grades)
    return "First Class Honours! Grade: " + str(result) if result >= 4.5 else "Second Class Noob! Grade: " + str(result)
```

```python
calculate(["A", "A-", "A-", "B+"])
```
